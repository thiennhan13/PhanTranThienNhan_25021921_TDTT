```python
import sys

def solve():
    input_data = sys.stdin.read().split()

    it = iter(map(int, input_data))
    
    try:
        t = next(it)
        
        for _ in range(t):
            k = next(it)
            n = next(it)
            
            k -= 1
            
            if n % 2 == 0:
                f = 1
            else:
                f = -1
            
            res = pow(k, n, 6971)
            
            ans = (res + f * k) % 6971
            
            print(ans, end=" ")
            
    except StopIteration:
        return

solve()
```
