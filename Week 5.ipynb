```python
#W5A3
def check(a : int):
    if(a < 2):
        return False
    r = int(a**0.5)
    for i in range (2, r + 1):
        if a % i == 0:
            return False
    return True
a= int(input().strip())
print(check(a))
```

```python
#W5A4
def check(a : int):
    if(a < 2):
        return False
    r = 0
    for i in range (1, a):
        if a % i == 0:
            r += i
    return (a == r)
a= int(input().strip())
print(check(a))
```

```python
#W5A6
def check(a : int):
    res = 1
    for i in range (1, a + 1):
        res *= i
    return res
a= int(input().strip())
print(check(a))
```

```python
#W5A7
def calc(num1, num2, op):
    if op == '+': return round(num1 + num2, 2)
    elif op == '-': return round(num1 - num2, 2)
    elif op == '*': return round(num1 * num2, 2)
    elif op == '/': return round(num1 / num2, 2)

a, b, c = input().split()
print(calc(float(a), float(b), c))
```

```python
#W5A8
import math
def Check(s1, s2):
    res = abs(len(s1) - len(s2))
    for i in range (0, min(len(s1), len(s2))):
        if(s1[i] != s2[i]):
            res += 1
    return res

a, b = map(int, input().split())
s1, s2 = '', ''

while(a > 0):
    s1 = str(a % 2) + s1
    a //= 2

while(b > 0):
    s2 = str(b % 2) + s2
    b //= 2
print(Check(s1, s2))
```

```python
#W5A10
def Check(a, b):
    char = {}
    if(len(a) != len(b)):
        return False
    for i in range (len(a)):
        if(a[i] in char):
            if(char[a[i]] != b[i]):
                return False
        else:
            char[a[i]] = b[i]
    return True
    
a = str(input())
b = str(input())
print(Check(a, b))
```
