```python
s = input()
x, y = int(input()), int(input())
kwh = y - x
cost = 0
if kwh <= 50:
    cost = kwh * 1984
elif kwh <= 100:
    cost = 50 * 1984 + (kwh - 50) * 2050
elif kwh <= 200:
    cost = 50 * 1984 + 50 * 2050 + (kwh - 100) * 2380
elif kwh <= 300:
    cost = 50 * 1984 + 50 * 2050 + 100 * 2380 + (kwh - 200) * 2998
elif kwh <= 400:
    cost = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (kwh - 300) * 3350
else:
    cost = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (kwh - 400) * 3460
print(f"Ho va ten: {s}")
print(f"Tien phai tra la: {int(cost * 1.08)}")
```
