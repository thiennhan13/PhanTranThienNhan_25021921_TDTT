```python
def Caesar(s, k):
    res = ""
    for ch in s:
        if(ch.isalpha()):
            if(ch.isupper()):
                start = ord('A')
            else:
                start = ord('a')
                
            new_char = chr((ord(ch) - start + k) % 26 + start)
            res += new_char

        else:
            if('0' <= ch <= '9'):
                start = ord('0')
                new_char = chr((ord(ch) - start + k) % 10 + start)
                res += new_char
            else:
                res += ch
    return res


s = input()
k = int(input())
print(Caesar(s, k))
```

```python
import sys

for line in sys.stdin:
    if not line.strip():
        continue

    try:
        n, k = map(int, line.split())
        res = n
        while(n >= k):
            new = n//k
            re = n % k
            res += new
            n = new + re
        print(res)
    except EOFError:
        break
```
